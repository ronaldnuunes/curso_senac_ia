```python
# Chamada 1 — Consultar CEP

import requests

url = "https://brasilapi.com.br/api/cep/v1/01001000"

response = requests.get(url)

print("Status:", response.status_code)
print(response.json())
```

```python
# Chamada 2 — Consultar DDD

url = "https://brasilapi.com.br/api/ddd/v1/11"

response = requests.get(url)

print("Status:", response.status_code)
print(response.json())
```

```python
# Chamada 3 — Consultar Bancos

url = "https://brasilapi.com.br/api/banks/v1"

response = requests.get(url)

print("Status:", response.status_code)

dados = response.json()

print(dados[:3])
```

```python
#4 Transformando JSON em DataFrame

import pandas as pd

df = pd.DataFrame(dados)

df.head()
```

```python
#Análise

print(df.shape)
print(df.columns)
```
